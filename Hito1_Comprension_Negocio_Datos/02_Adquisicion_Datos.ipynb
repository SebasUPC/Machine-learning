### 1.3.1 Recoleccion de datos

La funcion intenta descargar con `kagglehub`; si no hay conexion/credenciales, reutiliza el CSV local en `data/raw/`. Asi el pipeline es reproducible en cualquier entorno.

```python
import shutil, os
ac.DATA_RAW.mkdir(parents=True, exist_ok=True)

def obtener_dataset(kaggle_id, filename):
    destino = ac.DATA_RAW / filename
    try:
        import kagglehub
        ruta = kagglehub.dataset_download(kaggle_id)
        origen = os.path.join(ruta, filename)
        if os.path.exists(origen) and not destino.exists():
            shutil.copy(origen, destino)
        print(f'[kagglehub] {kaggle_id} -> {destino.name}')
    except Exception as e:
        if destino.exists():
            print(f'[fallback local] {destino.name} (kagglehub no disponible: {type(e).__name__})')
        else:
            raise FileNotFoundError(f'No se pudo obtener {filename}: {e}')
    return destino

ruta_salud = obtener_dataset(ac.KAGGLE_HEALTH_ID, ac.HEALTH_RAW_FILE)
ruta_stock = obtener_dataset(ac.KAGGLE_STOCK_ID, ac.STOCK_RAW_FILE)
```

### Verificacion de carga
